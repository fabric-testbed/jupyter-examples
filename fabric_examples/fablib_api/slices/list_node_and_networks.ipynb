## Background: Slice Components Hierarchy

A FABRIC slice is organized in a hierarchy of resources. Understanding this hierarchy helps you navigate and inspect your experiment's topology.


| Resource | Description | List Method |
|----------|-------------|-------------|
| **Node** | A virtual machine with CPU, RAM, and disk | `slice.list_nodes()` |
| **Network** | A Layer 2 or Layer 3 network link | `slice.list_networks()` |
| **Interface** | A network port on a node, connecting it to a network | `slice.list_interfaces()` |
| **Component** | Hardware attached to a node (NIC, GPU, FPGA, NVMe) | `slice.list_components()` |

All `list_*` methods support the same output options: default styled table, `output='text'`, `output='pandas'`, `output='json'`, `output='list'`. They also accept `fields` and `filter_function` parameters for customization.

## Step 1: Import FABlib and Verify Configuration

```python
# Import the FABlib library
from fabrictestbed_extensions.fablib.fablib import FablibManager as fablib_manager

# Create a FABlib manager instance
fablib = fablib_manager()

# Display current configuration
fablib.show_config();
```

---

## List Your Slices

First, list all slices to find a target slice with nodes and networks. For best results, use a slice that has multiple nodes and network connections.

```python
# List all slices -- find a name to use in the examples below
fablib.list_slices();
```

---

## Get a Slice by Name

Retrieve the slice object so we can inspect its internal components.

```python
# Set the target slice name -- replace with your actual slice name
slice_name='MySlice'

# Retrieve the slice object
slice = fablib.get_slice(name=slice_name)

# Show slice-level summary information
slice.show();
```

---

## List Nodes

The `slice.list_nodes()` method lists all **nodes** (virtual machines) in the slice, including their site, state, IP addresses, CPU/RAM/disk, and operating system.

Like all `list_*` methods, this supports `output`, `fields`, and `filter_function` parameters. See the [list_slices](./list_slices.ipynb) notebook for more information about output options.

```python
# List all nodes (VMs) in the slice
# Displays site, state, management IP, cores, RAM, disk, and image
output = slice.list_nodes()
```

---

## List Networks

The `slice.list_networks()` method lists all **networks** (Layer 2 and Layer 3 links) in the slice. This shows the network type, subnet, gateway, and which site hosts the network.

Like all `list_*` methods, this supports `output`, `fields`, and `filter_function` parameters. See the [list_slices](./list_slices.ipynb) notebook for more information about output options.

```python
# List all networks in the slice
# Displays network name, type, layer, subnet, and site
output = slice.list_networks()
```

---

## List Interfaces

The `slice.list_interfaces()` method lists all **interfaces** (network ports) across all nodes in the slice. Interfaces are the connection points between nodes and networks -- each interface belongs to a component on a node and connects to a network.

Like all `list_*` methods, this supports `output`, `fields`, and `filter_function` parameters. See the [list_slices](./list_slices.ipynb) notebook for more information about output options.

```python
# List all interfaces across all nodes in the slice
# Shows interface name, node, network, MAC address, and VLAN
output = slice.list_interfaces()
```

---

## List Components

The `slice.list_components()` method lists all **components** (hardware devices) attached to nodes in the slice. Components include NICs (Basic, ConnectX-5, ConnectX-6), GPUs (T4, RTX6000, A30, A40), FPGAs, and NVMe storage.

Like all `list_*` methods, this supports `output`, `fields`, and `filter_function` parameters. See the [list_slices](./list_slices.ipynb) notebook for more information about output options.

```python
# List all components across all nodes in the slice
# Shows component name, type/model, node, and details
output = slice.list_components()
```

## Troubleshooting

| Problem | Possible Cause | Solution |
|---------|---------------|----------|
| `Slice not found` | Slice name is misspelled or was deleted | Run `fablib.list_slices()` to verify available names |
| `list_nodes()` returns empty | Slice has no nodes (unlikely) or is still provisioning | Wait for slice state to reach `StableOK` |
| `list_networks()` returns empty | Slice has no network connections | This is normal for single-node slices without networking |
| `list_interfaces()` returns empty | No NICs or network components added | Add NIC components when creating the slice |
| `list_components()` returns empty | No additional components attached | This is normal for basic nodes with no GPUs, NICs, etc. |
| Output is hard to read | Too many columns displayed | Use `fields=['name','state']` to select specific columns |

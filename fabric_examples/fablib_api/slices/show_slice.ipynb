## Background: Getting vs. Listing Slices

FABlib provides two complementary approaches for working with slices:


Use `list_slices()` for a quick overview. Use `get_slice()` when you need to interact with a specific slice (show details, run commands, delete, renew, etc.).

The `slice.show()` method supports multiple output formats, similar to `list_slices()`:

| `output=` | Returns | Use Case |
|-----------|---------|----------|
| (default) | Styled Jupyter table | Interactive exploration |
| `'text'` | Plain text string | Logging, file output |
| `'json'` | JSON string | API integration |
| `'dict'` | Python dictionary | Programmatic access |

---

## Step 1: Import FABlib and Verify Configuration

```python
# Import the FABlib library
from fabrictestbed_extensions.fablib.fablib import FablibManager as fablib_manager

# Create a FABlib manager instance
fablib = fablib_manager()

# Display current configuration
fablib.show_config();
```

---

## List Your Slices First

Before inspecting a specific slice, list all your slices to find a target slice name and ID. Copy a name or ID from the output below to use in the remaining examples.

```python
# List all slices to find target slice names and IDs
fablib.list_slices();
```

---

## Get a Slice by Name

The most common way to retrieve a slice. Active slice names are unique per user, so this is unambiguous for current slices.

<div class="fab-warning">

**Tip:** Replace `'RCNF'` below with the name of one of your actual slices from the listing above.

</div>

```python
# Set the name of the slice to inspect
slice_name='RCNF'

# Retrieve the slice object by name
slice = fablib.get_slice(name=slice_name)

# Display all slice attributes as a styled table
slice.show();
```

### View Nodes and Networks in the Slice

Once you have a slice object, you can also list its nodes and networks.

```python
# List all nodes (VMs) in this slice
slice.list_nodes()

# List all networks (L2/L3 links) in this slice
slice.list_networks()
```

---

## Get a Slice by ID

Every slice has a unique **UUID** that is guaranteed to be unique across all slices -- including deleted and failed ones. Use the slice ID when you need to reference a specific historical slice.

<div class="fab-warning">

**Tip:** Replace `'VAILD_SLICE_ID'` below with an actual slice ID from the listing above.

</div>

```python
# Replace with valid slice ID
slice_id='VAILD_SLICE_ID'

# Attempt to retrieve the slice by ID
# Wrapped in try/except because the ID may not exist
try:
    slice = fablib.get_slice(slice_id=slice_id)
    slice.show();
except Exception as e:
    print(f"Slice ID '{slice_id}' not found, Exception: \n\n{e}")
```

---

## Get All Slices and Iterate

Use `fablib.get_slices()` to get a Python list of **Slice objects** for all your active slices. This lets you iterate and perform operations on each slice.

By default, this excludes `Dead` and `Closing` slices.

```python
# Iterate over all active slices and show their details
for slice in fablib.get_slices():
     slice.show()
```

---

## Select Specific Fields

Use the `fields` parameter to display only the columns you care about.

```python
slice_name='MySlice'

slice = fablib.get_slice(name=slice_name)

# Show only the name and state fields
slice.show(fields=['name','state']);
```

---

## Output as Text

Set `output='text'` to get a plain-text formatted string instead of a styled table. Useful for logging or saving to a file.

```python
slice_name='MySlice'

slice = fablib.get_slice(name=slice_name)

# Get slice details as a plain-text string
output_string = slice.show(output='text')
```

---

## Output as JSON String

Set `output='json'` to get slice details as a JSON-formatted string. Useful for API integration and structured data export.

```python
slice_name='MySlice'

slice = fablib.get_slice(name=slice_name)

# Get slice details as a JSON string
output_json_string = slice.show(output='json')
```

---

## Output as Python Dictionary

Set `output='dict'` to get a Python dictionary. This is the most flexible format for programmatic access to slice attributes.

```python
slice_name='MySlice'

slice = fablib.get_slice(name=slice_name)

# Get slice details as a Python dictionary
output_dict = slice.show(output='dict')
```

### Use the Dictionary Programmatically

Combine `output='dict'` with `quiet=True` to suppress display, then access specific fields by key.

```python
slice_name='MySlice'

slice = fablib.get_slice(name=slice_name)

# Get the dictionary without displaying anything (quiet=True)
output_dict = slice.show(output='dict', quiet=True)

# Access specific fields from the dictionary
print(f"{output_dict['name']}, {output_dict['state']}")
```

## Troubleshooting

| Problem | Possible Cause | Solution |
|---------|---------------|----------|
| `Slice not found` by name | Slice was deleted or name is misspelled | Run `fablib.list_slices()` to verify available names |
| `Slice not found` by ID | Invalid UUID or slice was purged | Double-check the ID string from a previous `list_slices()` call |
| `get_slices()` returns empty list | No active slices exist | Create a slice first using [Hello, FABRIC](../hello_fabric/hello_fabric.ipynb) |
| `show()` displays `None` for some fields | Slice is still being provisioned | Wait for the slice to reach `StableOK` state |
| `show_config()` shows missing values | Environment not configured | Run [Configure Environment](../../../configure_and_validate/configure_and_validate.ipynb) |

## Background: Why Delete Slices?

<div class="fab-danger">

**Important:** FABRIC is a shared research infrastructure. Always delete slices when you are finished with your experiment. Leaving unused slices running wastes resources and may prevent other researchers from getting the resources they need.

</div>

Slices have an **expiration time** (lease end date) after which they are automatically deleted. However, it is good practice to delete slices manually as soon as you are done.



---

## Step 1: Import FABlib and Verify Configuration

```python
# Import the FABlib library
from fabrictestbed_extensions.fablib.fablib import FablibManager as fablib_manager

# Create a FABlib manager instance
fablib = fablib_manager()

# Display current configuration to verify everything is set up correctly
fablib.show_config();
```

---

## Option 1: Get the Slice Object, Then Delete

This is the most explicit approach. You first retrieve the slice object using `fablib.get_slice()`, which lets you inspect it before deleting. This is useful when you want to confirm you are deleting the correct slice.

```python
# Set the name of the slice to delete
slice_name='MySlice'

# Retrieve the slice object by name
slice = fablib.get_slice(name=slice_name)

# Confirm which slice we are about to delete
print(f"Slice: {slice.get_name()}")

# Delete the slice and release all its resources
slice.delete()
```

---

## Option 2: Delete a Slice by Name

A shortcut that combines retrieval and deletion into one call. Pass the slice name directly to `fablib.delete_slice()`. This is the most concise approach when you know the exact slice name.

<div class="fab-warning">

**Tip:** Active slice names must be unique per user, so deleting by name is unambiguous for current slices. However, deleted or failed slices may share names -- use the slice ID for those cases.

</div>

```python
# Delete a slice directly by its name -- no need to get the slice object first
slice_name='MySlice'

fablib.delete_slice(slice_name)
```

## Troubleshooting

| Problem | Possible Cause | Solution |
|---------|---------------|----------|
| `Slice not found` exception | Slice was already deleted or name is misspelled | Run `fablib.list_slices()` to verify the slice name |
| `Slice not found` with ID | Invalid or expired slice ID | Double-check the ID string; it must be a valid UUID |
| Slice stuck in `Closing` state | FABRIC is still releasing resources | Wait a few minutes and check again with `fablib.list_slices()` |
| `delete_all()` misses some slices | Slices in error states may not be listed | Contact FABRIC support if slices persist after deletion |
| `PDP Authorization check failed` | Token expired or project permissions changed | Re-run the Configure Environment notebook to refresh tokens |

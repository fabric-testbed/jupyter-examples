# FABRIC Artifact Browser

The FABRIC [Artifact Manager](https://artifacts.fabric-testbed.net) is an invaluable tool for researchers and developers working with the FABRIC testbed. It facilitates the packaging, sharing, and reuse of complete, repeatable FABRIC experiments.

This notebook launches the **Artifact Browser**, an interactive tool for searching, filtering, and downloading artifacts directly from within your Jupyter environment.

**Features:**
- Browse the full artifact catalog with styled cards
- Filter by title or tag keywords in real time
- Select artifact versions from a dropdown
- One-click download with automatic tar extraction
- Scrollable interface that works well with 100+ artifacts

**Public Artifacts:** Any user can browse and download public artifacts.

**Project-Specific Artifacts:** Artifacts tied to specific projects are accessible by project members.

## Browse Artifacts

Run the cell below to launch the Artifact Browser. Make sure you have a valid FABRIC configuration (run the **Configure Environment** notebook first).

```python
from fabrictestbed_extensions.fablib.fablib import FablibManager
from fabrictestbed_extensions.fabvis import ArtifactBrowser

fablib = FablibManager()

browser = ArtifactBrowser(fablib)
browser.show()
```
